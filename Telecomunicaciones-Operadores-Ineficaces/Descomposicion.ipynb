## Caso principal – Telecomunicaciones: operadores ineficaces

### Objetivo de negocio

Ayudar al servicio de telefonía virtual **CallMeMaybe** a identificar a los operadores
menos eficaces. Un operador se considera ineficaz si:

- Tiene muchas llamadas entrantes perdidas.
- Tiene tiempos de espera prolongados para las llamadas entrantes.
- Realiza pocas llamadas salientes cuando se espera que haga más.

### Descomposición de tareas

1. **Comprender el contexto y los datos**
   - Leer la descripción del caso.
   - Revisar la estructura de los archivos `telecom_dataset_us.csv` y `telecom_clients_us.csv`
     (columnas, tipos de datos, tamaños).

2. **Carga y preparación de datos**
   - Cargar los datasets con `pandas`.
   - Convertir columnas de fecha a tipo `datetime`.
   - Revisar valores ausentes y duplicados, describir su impacto y decidir cómo tratarlos.
   - Crear, si es necesario, columnas derivadas (por ejemplo indicadores de llamadas internas/externas).

3. **Análisis exploratorio de datos (EDA)**
   - Analizar el volumen de llamadas por tipo (entrantes, salientes, internas, externas).
   - Estudiar la distribución de la duración y el número de llamadas.
   - Detectar patrones por fecha, tipo de llamada u operador.

4. **Definir métricas de eficacia / ineficacia**
   - Calcular métricas a nivel operador:
     - tasa de llamadas perdidas,
     - número de llamadas salientes,
     - duración media de las llamadas.
   - Definir reglas claras para marcar a un operador como “ineficaz”
     (por ejemplo umbrales basados en percentiles).

5. **Identificar operadores ineficaces**
   - Construir una tabla resumen por `operator_id` con todas las métricas.
   - Marcar a los operadores que cumplan los criterios de ineficacia.
   - Analizar cuántos operadores son ineficaces y qué características comparten.

6. **Pruebas de hipótesis**
   - Formular 1–2 hipótesis estadísticas relacionadas con la ineficacia
     (por ejemplo, comparar tasa de llamadas perdidas entre operadores ineficaces y el resto).
   - Elegir el tipo de prueba adecuado (proporciones o medias).
   - Calcular los resultados y extraer conclusiones.

7. **Dashboard y presentación**
   - Preparar un dataset limpio para Tableau (nivel llamada y/o nivel operador).
   - Crear un dashboard que muestre:
     - distribución de duración de llamadas
     - proporción de llamadas internas vs externas
     - métricas clave por operador
   - Elaborar una presentación (PDF) con:
     - contexto del caso
     - enfoque analítico
     - resultados principales
     - lista de operadores ineficaces
     - recomendaciones para el negocio

## Caso 2 – Test A/B del sistema de recomendación

### Objetivo de negocio

Evaluar si el nuevo sistema de recomendaciones mejora la conversión en el embudo:

`product_page -> product_card -> purchase`.

Se debe analizar la prueba `recommender_system_test` y decidir
si el nuevo sistema debe implementarse.

### Descomposición de tareas

1. **Comprender la especificación de la prueba**
   - Revisar la descripción técnica: fechas, tamaño esperado de la muestra,
     grupos A (control) y B (nuevo sistema), eventos clave del embudo.

2. **Carga y preparación de datos**
   - Cargar los cuatro archivos CSV del experimento.
   - Convertir fechas a tipo `datetime`.
   - Revisar y describir valores ausentes y duplicados.

3. **Verificación de la calidad de la prueba**
   - Comprobar que cada usuario pertenece solo a un grupo.
   - Revisar la ventana temporal de inscripción y eventos.
   - Analizar si el tamaño final de las muestras es razonable.

4. **Análisis exploratorio y construcción del embudo**
   - Calcular el número de eventos por usuario y por día.
   - Construir el embudo para cada grupo:
     - usuarios que visitan la página de producto
     - usuarios que agregan al carrito
     - usuarios que realizan una compra
   - Calcular las tasas de conversión en cada paso y a lo largo del embudo completo

5. **Pruebas de hipótesis**
   - Formular hipótesis sobre la diferencia de conversión entre los grupos A y B.
   - Utilizar pruebas estadísticas (por ejemplo, prueba z para proporciones).
   - Evaluar la significancia estadística y la magnitud del efecto.

6. **Conclusiones y decisión**
   - Resumir los resultados del experimento.
   - Indicar si se alcanza el aumento esperado (≥10% de mejora).
   - Tomar una decisión:
     - declarar ganador a un grupo
     - concluir que no hay diferencias
     - o recomendar repetir/continuar la prueba

## Caso 3 – SQL: servicio de libros

### Objetivo de negocio

Analizar la base de datos de un servicio de libros (libros, autores,
editoriales, reseñas y calificaciones) para obtener insights que apoyen
el diseño de una nueva propuesta de valor.

### Descomposición de tareas

1. **Conexión a la base de datos**
   - Configurar la conexión a PostgreSQL utilizando `sqlalchemy` y el certificado `CA.pem`.
   - Probar una consulta simple a la tabla `books`.

2. **Exploración de las tablas**
   - Inspeccionar las tablas `books`, `authors`, `publishers`,
     `ratings` y `reviews` (primeras filas y estructura).
   - Entender las claves primarias y foráneas usando el diagrama de datos.

3. **Consultas SQL principales**
   - Libros publicados después del 1 de enero del 2000.
   - Número de reseñas y rating promedio para cada libro.
   - Editorial con más libros de más de 50 páginas.
   - Autor con la calificación promedio más alta considerando
     solo libros con al menos 50 calificaciones.
   - Número promedio de reseñas de texto entre usuarios que calificaron más de 50 libros.

4. **Resultados y conclusiones**
   - Mostrar los resultados de cada consulta en el Notebook.
   - Redactar una breve conclusión para cada tarea.
   - Resumir qué tipo de libros, editoriales y autores destacan
     según el análisis, y cómo esto puede ayudar a definir el producto.
